## El Problema: BigCompany Inc. Se Ahoga en Tickets de Soporte

**BigCompany Inc.** es una empresa SaaS de rápido crecimiento que vende una plataforma de gestión de proyectos llamada **BigProject Pro**. Su producto es popular — pero su equipo de atención al cliente tiene un problema serio.

Cada día, llegan cientos de tickets de soporte. Algunos son preguntas sencillas que podrían responderse en segundos ("¿Cómo restablezco mi contraseña?"). Otros son problemas complejos que requieren una investigación cuidadosa ("Me han cobrado dos veces y necesito un reembolso"). Pero **todos los tickets van a la misma cola**, y cada ticket recibe el mismo tratamiento:

- Reciben **más de 400 tickets al día**.
- Los agentes de soporte dedican **15 minutos** a preguntas sencillas de FAQ que deberían llevar 30 segundos.
- Los **problemas complejos de facturación y técnicos** esperan en la cola detrás de restablecimientos de contraseña.
- El tiempo medio de primera respuesta se ha disparado a **4,5 horas**.
- La satisfacción del cliente (CSAT) ha caído al **62%** — muy por debajo de la media del sector del 78%.
- El equipo está **agotado** de responder las mismas preguntas sencillas cientos de veces.

La Directora de Atención al Cliente convoca una reunión de emergencia:

> *"Necesitamos un sistema que pueda averiguar automáticamente qué tipo de ticket es, responder los sencillos al instante, y derivar los complejos a un equipo de investigación completo. Nuestros agentes solo deberían ocuparse de tickets que realmente necesiten un cerebro humano."*

El equipo de ingeniería propone construir un **sistema de atención al cliente potenciado por IA** usando **CrewAI Flows + Crews**.

## ¿Por Qué CrewAI? Y Qué Hay de Nuevo Aquí — ¡Flows!

En los notebooks anteriores, aprendimos a construir **Crews** — equipos de agentes de IA que colaboran en tareas. Un Crew es genial para resolver un solo problema (como puntuar un lead o escribir contenido).

Pero la atención al cliente tiene un desafío que un solo Crew no puede manejar bien: **el enrutamiento**.

No todos los tickets necesitan el mismo tratamiento:
- Una **pregunta sencilla** ("¿Cuáles son vuestros precios?") debería ser respondida por un solo agente usando las FAQ.
- Un **problema complejo** ("Me han cobrado dos veces 450 €") debería ir a un equipo de investigación multi-agente completo.

Aquí es donde entran los **Flows de CrewAI**.

### ¿Qué Es un Flow?

Un **Flow** es un flujo de trabajo estructurado que orquesta *cuándo* y *cómo* suceden las cosas. Piensa en él como el **gerente** que decide la secuencia de operaciones:

| Concepto | Qué Hace | Analogía |
|----------|----------|----------|
| **Crew** | Un equipo de agentes que colaboran en una tarea compartida | Un equipo de trabajadores en un departamento |
| **Flow** | Un orquestador que dirige el trabajo al crew correcto | El jefe de planta que asigna tickets al departamento adecuado |

### Decoradores de Flow — El Concepto Clave

Los Flows usan **decoradores de Python** — anotaciones especiales que definen cuándo y cómo se ejecuta cada paso:

| Decorador | Qué Hace | Ejemplo |
|-----------|----------|----------|
| `@start()` | Marca el punto de entrada del flow | Recibir un nuevo ticket de soporte |
| `@listen(method)` | Se ejecuta cuando un método específico se completa | Después de que el ticket sea clasificado |
| `@router(method)` | Enruta a diferentes caminos según una condición | Ticket sencillo → vía rápida; complejo → crew completo |
| `or_(a, b)` | Se dispara cuando **cualquiera** de los métodos a o b se completa | Después de la vía que haya gestionado el ticket |

### La Visión General: Crews Dentro de Flows

El verdadero poder viene de **combinar Crews y Flows**:

```
Flow (el orquestador)
  ├── @start()  →  Clasificar el ticket
  ├── @router() →  Enrutar según categoría
  │     ├── "simple"  →  Crew de un solo agente (búsqueda en FAQ)
  │     └── "complex" →  Crew multi-agente (Investigación + Solución)
  └── @listen()  →  Formatear y entregar la respuesta
```

Esta es la **arquitectura dual** de CrewAI — los Flows manejan la lógica de enrutamiento, los Crews manejan la resolución colaborativa de problemas.

## La Solución: Cómo Funciona el Sistema de Soporte con IA

Este es el flujo de trabajo que diseñó BigCompany Inc.:


### Paso 3: Definir las Herramientas del Agente

Las **herramientas** son funciones que los agentes pueden invocar para buscar información. Crearemos dos herramientas:

| Herramienta | Qué Hace | Usada Por |
|-------------|----------|----------|
| `search_faq` | Busca en la base de datos de FAQ respuestas coincidentes | Agente FAQ (vía sencilla) |
| `search_knowledge_base` | Busca en la documentación detallada | Agente Investigador (vía compleja) |

El decorador `@tool` convierte una función regular de Python en una herramienta que los agentes pueden usar. El **docstring** es crítico — el LLM lo lee para decidir *cuándo* y *cómo* usar cada herramienta.

````python
from crewai.tools import tool


@tool("search_faq")
def search_faq(query: str) -> str:
    """Search the FAQ database for answers to common customer questions.

    Use this tool to find pre-written answers for simple questions about
    password resets, pricing, cancellations, and integrations.

    Args:
        query: A keyword or topic to search for (e.g., 'password', 'pricing').
    """
    query_lower = query.lower()
    results = []

    for key, entry in FAQ_DB.items():
        if (
            query_lower in key
            or query_lower in entry["question"].lower()
            or query_lower in entry["answer"].lower()
        ):
            results.append(f"Q: {entry['question']}\nA: {entry['answer']}")

    if results:
        return "\n\n---\n\n".join(results)
    return f"No FAQ entries found for '{query}'. Try different keywords."


@tool("search_knowledge_base")
def search_knowledge_base(topic: str) -> str:
    """Search the detailed knowledge base for policies and procedures.

    Use this tool to find in-depth information about billing policies,
    technical troubleshooting steps, or account management procedures.

    Args:
        topic: The topic to search for (e.g., 'billing', 'technical', 'account').
    """
    topic_lower = topic.lower()
    results = []

    for key, entry in KNOWLEDGE_BASE.items():
        if (
            topic_lower in key
            or topic_lower in entry["topic"].lower()
            or topic_lower in entry["content"].lower()
        ):
            results.append(f"Topic: {entry['topic']}\n{entry['content']}")

    if results:
        return "\n\n---\n\n".join(results)
    return f"No knowledge base entries found for '{topic}'. Try different keywords."


print("Herramientas creadas:")
print(f"  - search_faq: Busca en {len(FAQ_DB)} entradas de FAQ")
print(f"  - search_knowledge_base: Busca en {len(KNOWLEDGE_BASE)} temas de la base de conocimiento")
````

### Paso 4: Crear los Agentes

Necesitamos cuatro agentes — cada uno con un rol específico en el pipeline de soporte:

| Agente | Rol | Usado En | Herramientas |
|--------|-----|----------|-------------|
| **Clasificador** | Categoriza tickets por tipo y urgencia | Flow (paso de clasificación) | Ninguna |
| **Especialista FAQ** | Responde preguntas sencillas desde las FAQ | Crew de la vía sencilla | `search_faq` |
| **Investigador de Soporte** | Investiga problemas complejos | Crew de la vía compleja | `search_knowledge_base` |
| **Redactor de Soluciones** | Escribe respuestas detalladas | Crew de la vía compleja | Ninguna |

Observa cómo:
- El **Clasificador** y el **Especialista FAQ** trabajan solos (crews de un solo agente).
- El **Investigador** y el **Redactor de Soluciones** trabajan juntos (crew multi-agente para problemas complejos).

#### Nota de actualización: modelos actuales de OpenAI
* El ejemplo original es muy bueno para aprender el concepto, y funcionaba con el modelo de OpenAI con el que se grabó: `gpt-4o-mini`, con `temperature=0`.
* OpenAI va a retirar `gpt-4o-mini` (retirada prevista el 14 de enero de 2027). Con los modelos actuales, usaríamos `gpt-6-luna`, el modelo pequeño que recomienda OpenAI ahora. `MODEL` lo define una sola vez, para que puedas cambiarlo tocando una línea.
* Los modelos recientes de OpenAI solo aceptan la temperatura por defecto (1), así que el código usa `temperature=1`. La idea de `temperature=0` (respuestas más predecibles, útiles para clasificar los tickets siempre igual) sigue siendo importante, y otros modelos sí permiten elegirla.
* Los agentes de este flow usan herramientas. Por la vía que usa CrewAI, `gpt-6-luna` solo acepta herramientas con el razonamiento desactivado, así que el código añade `extra_body={"reasoning_effort": "none"}`. Así el modelo responde directamente, como hacía `gpt-4o-mini`. Funciona con la versión de CrewAI del repositorio del backend (1.9.3) y con la más reciente, la que instala `pip install -U crewai`.
* Consulta los modelos disponibles en developers.openai.com/api/docs/deprecations.

````python
from crewai import Agent, LLM

# Configurar el LLM (el "cerebro" de todos los agentes)
# Versión original: LLM(model="openai/gpt-4o-mini", temperature=0).
# OpenAI va a retirar gpt-4o-mini (retirada prevista el 14 de enero de 2027),
# así que usamos gpt-6-luna, el modelo pequeño que recomienda OpenAI ahora.
# Consulta los modelos disponibles en developers.openai.com/api/docs/deprecations
MODEL = "openai/gpt-6-luna"

# Los modelos recientes de OpenAI solo aceptan la temperatura por defecto (1),
# así que usamos 1 en lugar de 0. Los agentes de este flow usan herramientas y,
# por la vía que usa CrewAI (Chat Completions), gpt-6-luna solo acepta
# herramientas con el razonamiento desactivado, así que fijamos
# reasoning_effort en "none". Lo enviamos con extra_body, que funciona con la
# versión de CrewAI de este proyecto y con la más reciente.
llm = LLM(model=MODEL, temperature=1, extra_body={"reasoning_effort": "none"})

# ── Agente 1: Clasificador de Tickets ─────────────────────
classifier_agent = Agent(
    role="Customer Support Ticket Classifier",
    goal=(
        "Accurately classify incoming support tickets by category "
        "and urgency level so they can be routed to the right team"
    ),
    backstory=(
        "You are a senior support operations specialist with 8 years "
        "of experience in customer support triage. You can quickly "
        "identify whether a ticket is a simple FAQ question or a "
        "complex issue requiring investigation. You never misclassify "
        "a billing dispute as a simple question."
    ),
    llm=llm,
    verbose=True,
    allow_delegation=False,
)

# ── Agente 2: Especialista FAQ ─────────────────────────────
faq_agent = Agent(
    role="FAQ Support Specialist",
    goal=(
        "Quickly answer common customer questions using the FAQ "
        "database, providing clear and helpful responses"
    ),
    backstory=(
        "You are a friendly and efficient frontline support agent. "
        "You know the FAQ inside and out and can answer common "
        "questions in a warm, helpful tone. You always point "
        "customers to the exact steps they need to follow."
    ),
    llm=llm,
    tools=[search_faq],
    verbose=True,
    allow_delegation=False,
)

# ── Agente 3: Investigador de Soporte ─────────────────────
research_agent = Agent(
    role="Support Research Specialist",
    goal=(
        "Thoroughly investigate complex customer issues by searching "
        "the knowledge base for relevant policies and procedures"
    ),
    backstory=(
        "You are a detail-oriented support investigator who leaves "
        "no stone unturned. When a customer has a billing dispute "
        "or technical problem, you dig into the knowledge base to "
        "find every relevant policy, procedure, and troubleshooting "
        "step before passing your findings to the solution team."
    ),
    llm=llm,
    tools=[search_knowledge_base],
    verbose=True,
    allow_delegation=False,
)

# ── Agente 4: Redactor de Soluciones ──────────────────────
solution_agent = Agent(
    role="Customer Solution Specialist",
    goal=(
        "Write a clear, empathetic, and actionable response to the "
        "customer based on the research findings"
    ),
    backstory=(
        "You are a senior support agent known for turning frustrated "
        "customers into happy ones. You write responses that are "
        "empathetic, clear, and action-oriented. You always "
        "acknowledge the customer's frustration, explain what "
        "happened, and provide concrete next steps."
    ),
    llm=llm,
    verbose=True,
    allow_delegation=False,
)

print("¡Equipo de soporte creado!")
print(f"  - Agente 1: {classifier_agent.role}")
print(f"  - Agente 2: {faq_agent.role}")
print(f"  - Agente 3: {research_agent.role}")
print(f"  - Agente 4: {solution_agent.role}")
````

### Paso 5: Definir los Modelos de Salida Estructurada

Al igual que en los casos de uso anteriores, usamos **modelos Pydantic** para asegurar que nuestros agentes devuelvan datos en un formato consistente y estructurado.

Definimos dos "formularios" de salida:
- **TicketClassification** — La salida del clasificador (categoría, complejidad, urgencia).
- **SupportResponse** — La respuesta final al cliente.

````python
from pydantic import BaseModel, Field
from typing import List


# ── Formato de salida para el Agente Clasificador ─────────
class TicketClassification(BaseModel):
    """Structured output from the Ticket Classifier."""

    category: str = Field(
        description=(
            "Ticket category: 'password_reset', 'pricing', "
            "'cancel_subscription', 'integrations', 'billing_issue', "
            "'technical_issue', or 'account_issue'"
        )
    )
    complexity: str = Field(
        description="'simple' (FAQ can answer) or 'complex' (needs investigation)"
    )
    urgency: str = Field(
        description="'low', 'medium', or 'high'"
    )
    reasoning: str = Field(
        description="Brief explanation of why this classification was chosen"
    )


# ── Formato de salida para la Respuesta de Soporte ────────
class SupportResponse(BaseModel):
    """Structured response to send to the customer."""

    subject: str = Field(description="Response subject line")
    message: str = Field(description="The full response message to the customer")
    follow_up_needed: bool = Field(
        description="Whether a human agent should follow up"
    )


print("Modelos de salida estructurada definidos:")
print("  - TicketClassification: 4 campos (category, complexity, urgency, reasoning)")
print("  - SupportResponse: 3 campos (subject, message, follow_up_needed)")
````

### Paso 6: Definir el Estado del Flow

Un **Flow** necesita un lugar para almacenar datos a medida que avanza por cada paso. Esto se llama el **Estado del Flow**.

Definimos un modelo Pydantic que contiene todo sobre el ticket mientras se procesa:

```
Paso 1 (clasificar): rellena → category, complexity, urgency
Paso 2 (enrutar):    lee     → complexity (para decidir el camino)
Paso 3 (gestionar):  rellena → resolution
Paso 4 (formatear):  lee     → todo, devuelve la salida final
```

El estado se comparte entre todos los métodos del flow a través de `self.state`.

````python
class SupportTicketState(BaseModel):
    """State that flows through the entire support pipeline."""

    # Campos de entrada (se establecen antes del kickoff)
    customer_name: str = ""
    customer_email: str = ""
    ticket_subject: str = ""
    ticket_message: str = ""

    # Campos de clasificación (establecidos por classify_ticket)
    category: str = ""
    complexity: str = ""  # "simple" o "complex"
    urgency: str = ""

    # Campos de resolución (establecidos por handle_simple o handle_complex)
    resolution_subject: str = ""
    resolution_message: str = ""
    follow_up_needed: bool = False
    path_taken: str = ""  # "simple" o "complex" — para seguimiento


print("Modelo de estado del flow definido: SupportTicketState")
print("  Campos de entrada: customer_name, customer_email, ticket_subject, ticket_message")
print("  Campos de clasificación: category, complexity, urgency")
print("  Campos de resolución: resolution_subject, resolution_message, follow_up_needed")
````

### Paso 7: Construir el Flow — El Corazón de la Aplicación

Ahora construimos el **CustomerSupportFlow** — el orquestador que lo conecta todo.

Aquí es donde entran los nuevos conceptos:

1. **`@start()`** — El punto de entrada. Cuando llamamos a `flow.kickoff()`, este método se ejecuta primero. Crea un Crew de un solo agente para clasificar el ticket.

2. **`@router(classify_ticket)`** — Después de la clasificación, este método examina `self.state.complexity` y devuelve `"simple"` o `"complex"`. Esta cadena se convierte en una **etiqueta de ruta**.

3. **`@listen("simple")`** — Este método solo se ejecuta si el router devolvió `"simple"`. Crea un Crew FAQ de un solo agente.

4. **`@listen("complex")`** — Este método solo se ejecuta si el router devolvió `"complex"`. Crea un Crew multi-agente con un Investigador y un Redactor de Soluciones.

5. **`@listen(or_(handle_simple_ticket, handle_complex_ticket))`** — Este método se ejecuta después de que **cualquiera** de las vías se complete. `or_()` significa "disparar cuando cualquiera de estos métodos termine". Formatea la salida final.

Veamos el código:

````python
from crewai import Task, Crew, Process
from crewai.flow.flow import Flow, start, listen, router, or_


class CustomerSupportFlow(Flow[SupportTicketState]):
    """
    Flow de atención al cliente potenciado por IA de BigCompany Inc.

    Este Flow orquesta todo el pipeline de soporte:
    1. Clasificar el ticket (categoría + complejidad)
    2. Enrutar al camino correcto (FAQ sencilla o investigación compleja)
    3. Resolver el ticket usando el Crew apropiado
    4. Formatear y entregar la respuesta final
    """

    # ── Paso 1: Clasificar el Ticket ──────────────────────────
    @start()
    def classify_ticket(self):
        """Punto de entrada: clasificar el ticket entrante."""
        print(f"\n{'='*60}")
        print(f"NUEVO TICKET de {self.state.customer_name}")
        print(f"Asunto: {self.state.ticket_subject}")
        print(f"{'='*60}")

        # Crear una tarea de clasificación
        classify_task = Task(
            description=(
                f"Classify this customer support ticket:\n"
                f"  Customer: {self.state.customer_name}\n"
                f"  Subject: {self.state.ticket_subject}\n"
                f"  Message: {self.state.ticket_message}\n\n"
                f"Determine:\n"
                f"1. Category: password_reset, pricing, cancel_subscription, "
                f"integrations, billing_issue, technical_issue, or account_issue\n"
                f"2. Complexity: 'simple' if it's a common FAQ question, "
                f"'complex' if it requires investigation (billing disputes, "
                f"technical bugs, account problems)\n"
                f"3. Urgency: 'low', 'medium', or 'high'\n"
                f"   (billing disputes and data loss = high, "
                f"feature questions = low)"
            ),
            expected_output=(
                "A classification with category, complexity, urgency, "
                "and reasoning."
            ),
            agent=classifier_agent,
            output_pydantic=TicketClassification,
        )

        # Ejecutar un crew de un solo agente para la clasificación
        classify_crew = Crew(
            agents=[classifier_agent],
            tasks=[classify_task],
            process=Process.sequential,
            verbose=True,
        )
        classify_crew.kickoff()

        # Almacenar los resultados de clasificación en el estado del flow
        if classify_task.output.pydantic:
            classification = classify_task.output.pydantic
            self.state.category = classification.category
            self.state.complexity = classification.complexity
            self.state.urgency = classification.urgency

        print(f"\nClasificación: {self.state.category} | "
              f"{self.state.complexity} | urgencia={self.state.urgency}")

        return self.state.complexity

    # ── Paso 2: Enrutar el Ticket ─────────────────────────────
    @router(classify_ticket)
    def route_ticket(self):
        """
        Router: decide qué camino toma el ticket.

        Devuelve una cadena — "simple" o "complex".
        Los métodos decorados con @listen("simple") o @listen("complex")
        se dispararán según este valor de retorno.
        """
        if self.state.complexity == "simple":
            print("\n>> ENRUTAMIENTO: Vía sencilla (agente FAQ)")
            return "simple"
        else:
            print("\n>> ENRUTAMIENTO: Vía compleja (crew de investigación)")
            return "complex"

    # ── Paso 3a: Gestionar Tickets Sencillos (Vía FAQ) ────────
    @listen("simple")
    def handle_simple_ticket(self):
        """
        Se ejecuta solo cuando route_ticket() devuelve "simple".
        Usa un solo agente FAQ para responder la pregunta.
        """
        print("\n--- Vía Sencilla: Agente FAQ gestionando ticket ---")

        faq_task = Task(
            description=(
                f"Answer this customer support question using the FAQ database:\n"
                f"  Customer: {self.state.customer_name}\n"
                f"  Question: {self.state.ticket_message}\n\n"
                f"Search the FAQ for relevant information and write a friendly, "
                f"helpful response. Address the customer by name."
            ),
            expected_output="A helpful response with clear instructions.",
            agent=faq_agent,
            output_pydantic=SupportResponse,
        )

        faq_crew = Crew(
            agents=[faq_agent],
            tasks=[faq_task],
            process=Process.sequential,
            verbose=True,
        )
        faq_crew.kickoff()

        # Almacenar la resolución en el estado del flow
        if faq_task.output.pydantic:
            response = faq_task.output.pydantic
            self.state.resolution_subject = response.subject
            self.state.resolution_message = response.message
            self.state.follow_up_needed = response.follow_up_needed

        self.state.path_taken = "simple"
        return self.state.resolution_message

    # ── Paso 3b: Gestionar Tickets Complejos (Vía de Investigación) ──
    @listen("complex")
    def handle_complex_ticket(self):
        """
        Se ejecuta solo cuando route_ticket() devuelve "complex".
        Usa un crew multi-agente: Investigador + Redactor de Soluciones.
        """
        print("\n--- Vía Compleja: Crew de Investigación gestionando ticket ---")

        # Tarea 1: Investigar el problema
        research_task = Task(
            description=(
                f"Investigate this customer support issue:\n"
                f"  Customer: {self.state.customer_name}\n"
                f"  Category: {self.state.category}\n"
                f"  Issue: {self.state.ticket_message}\n\n"
                f"Search the knowledge base for relevant policies, procedures, "
                f"and troubleshooting steps. Compile all relevant findings."
            ),
            expected_output=(
                "A detailed research report with relevant policies, "
                "procedures, and recommended actions."
            ),
            agent=research_agent,
        )

        # Tarea 2: Redactar la respuesta de solución
        solution_task = Task(
            description=(
                f"Based on the research findings, write a response to "
                f"{self.state.customer_name} about their issue:\n"
                f"  Issue: {self.state.ticket_message}\n\n"
                f"Guidelines:\n"
                f"- Acknowledge their frustration\n"
                f"- Explain what happened (if applicable)\n"
                f"- Provide clear next steps\n"
                f"- Be empathetic and professional\n"
                f"- Set follow_up_needed=True if a human should follow up"
            ),
            expected_output=(
                "A professional, empathetic customer response with "
                "clear next steps."
            ),
            agent=solution_agent,
            context=[research_task],
            output_pydantic=SupportResponse,
        )

        # Ejecutar el crew de investigación multi-agente
        investigation_crew = Crew(
            agents=[research_agent, solution_agent],
            tasks=[research_task, solution_task],
            process=Process.sequential,
            verbose=True,
        )
        investigation_crew.kickoff()

        # Almacenar la resolución en el estado del flow
        if solution_task.output.pydantic:
            response = solution_task.output.pydantic
            self.state.resolution_subject = response.subject
            self.state.resolution_message = response.message
            self.state.follow_up_needed = response.follow_up_needed

        self.state.path_taken = "complex"
        return self.state.resolution_message

    # ── Paso 4: Formatear la Respuesta Final ──────────────────
    @listen(or_(handle_simple_ticket, handle_complex_ticket))
    def format_final_response(self, resolution):
        """
        Se ejecuta después de que se complete la vía sencilla O la compleja.

        or_() significa: disparar cuando CUALQUIERA de estos métodos termine.
        Como solo se ejecuta una vía por ticket (gracias al router),
        esto efectivamente significa "ejecutar después de la vía que se haya tomado".
        """
        print(f"\n--- Formateando respuesta final (vía: {self.state.path_taken}) ---")

        # Construir la respuesta final formateada
        final = (
            f"Para: {self.state.customer_name} <{self.state.customer_email}>\n"
            f"Asunto: {self.state.resolution_subject}\n"
            f"Vía: {self.state.path_taken.upper()} | "
            f"Categoría: {self.state.category} | "
            f"Urgencia: {self.state.urgency}\n"
            f"Seguimiento necesario: {'Sí' if self.state.follow_up_needed else 'No'}\n"
            f"{'='*60}\n\n"
            f"{self.state.resolution_message}"
        )

        return final


print("¡CustomerSupportFlow definido!")
print("  Pasos:")
print("    1. @start()              → classify_ticket")
print("    2. @router()             → route_ticket (devuelve 'simple' o 'complex')")
print("    3a. @listen('simple')    → handle_simple_ticket (crew FAQ)")
print("    3b. @listen('complex')   → handle_complex_ticket (crew de investigación)")
print("    4. @listen(or_(...))     → format_final_response")
````

```

│                                                              │

│                       │                                      │

│          │                           │                       │
│  ┌───────▼──────────┐    ┌───────────▼──────────────┐        │
│  │ @listen("simple")     │ @listen("complex")       │        │
│  │                  │    │                          │        │
│  │ Crew de un solo  │    │ Crew multi-agente:       │        │
│  │ agente:          │    │  - Agente Investigador   │        │
│  │  - Agente FAQ    │    │  - Redactor Soluciones   │        │
│  │  - herramienta   │    │  - herramienta           │        │
│  │    search_faq    │    │    search_knowledge_base │        │
│  └───────┬──────────┘    └───────────┬──────────────┘        │
│          │                           │                       │
│  ┌───────▼───────────────────────────▼──────────────────┐    │
│  │  @listen(or_(...)) format_final_response             │    │
│  │  → Formatea y entrega la respuesta                   │    │
│  └──────────────────────────────────────────────────────┘    │
└──────────────────────────────────────────────────────────────┘
```

### Componentes Clave

| Componente | Qué Es | En Nuestro Código |
|------------|--------|------------------|
| **Flow** | El orquestador que enruta tickets | `CustomerSupportFlow` |
| **Estado del Flow** | Datos compartidos entre todos los pasos | `SupportTicketState` (modelo Pydantic) |
| **`@start()`** | Punto de entrada del flow | `classify_ticket()` |
| **`@router()`** | Bifurcación condicional | `route_ticket()` → devuelve `"simple"` o `"complex"` |
| **`@listen("ruta")`** | Escucha una ruta específica | `handle_simple_ticket()`, `handle_complex_ticket()` |
| **`or_()`** | Se dispara cuando cualquier método se completa | `format_final_response()` se ejecuta después de cualquier vía |
| **Crew** | Equipo de agentes para una tarea específica | Crew FAQ (1 agente), Crew de investigación (2 agentes) |
| **Agent** | Un miembro del equipo IA con un rol | Clasificador, Especialista FAQ, Investigador, Redactor de Soluciones |
| **Task** | Una asignación específica para un agente | Clasificación, búsqueda FAQ, Investigación, Redacción de solución |
| **Tool** | Una función que los agentes pueden invocar | `search_faq`, `search_knowledge_base` |

## Conclusiones Clave para Principiantes

1. **Los Flows son el orquestador; los Crews son los trabajadores.** Un Flow decide *qué* necesita suceder y *cuándo*. Un Crew decide *cómo* hacerlo. Juntos, crean aplicaciones de IA potentes y estructuradas.

2. **`@start()` es el punto de entrada.** Cuando llamas a `flow.kickoff()`, todos los métodos decorados con `@start()` se ejecutan primero.

3. **`@router()` habilita el enrutamiento inteligente.** El método router devuelve una cadena (como `"simple"` o `"complex"`), y los métodos `@listen()` coinciden con esa cadena. Así es como creas diferentes caminos para diferentes situaciones.

4. **`@listen()` puede escuchar métodos o nombres de ruta.** `@listen(classify_ticket)` se dispara cuando ese método se completa. `@listen("simple")` se dispara cuando un router devuelve esa cadena.

5. **`or_()` combina múltiples disparadores.** `@listen(or_(method_a, method_b))` se dispara cuando *cualquiera* de los métodos se completa — perfecto para fusionar vías paralelas de vuelta.

6. **El Estado del Flow comparte datos entre pasos.** El modelo Pydantic `SupportTicketState` fluye a través de cada paso. Cada método puede leer y escribir en `self.state`.

7. **Puedes llamar a `crew.kickoff()` dentro de cualquier método del flow.** Así es como combinas el poder de enrutamiento de los Flows con la inteligencia colaborativa de los Crews.

8. **Cada prueba obtiene una instancia fresca del flow.** Creamos un nuevo `CustomerSupportFlow()` para cada ticket. El estado del flow no se comparte entre tickets.

## ¿Qué Viene Después?

Este es un sistema de atención al cliente sencillo pero funcional. En un entorno de producción en BigCompany Inc., el equipo añadiría:

- **Decorador `@human_feedback()`** — El mecanismo integrado de CrewAI para flujos de trabajo con intervención humana. Cuando la IA no está segura, el flow puede pausarse y presentar la respuesta redactada a un agente humano para su aprobación antes de enviarla al cliente. Se importa con `from crewai.flow.human_feedback import human_feedback`.

- **Decorador `@persist`** — Guarda el estado del flow en una base de datos SQLite después de cada paso. Si el sistema se reinicia o un agente humano tarda horas en revisar una escalación, el flujo de trabajo continúa exactamente donde se quedó. Se importa con `from crewai.flow.persistence import persist`.

- **Integración real con base de conocimiento mediante MCP** — Usar el soporte de Model Context Protocol (MCP) de CrewAI para conectarse directamente a Zendesk, Confluence u otros sistemas de gestión del conocimiento en lugar de diccionarios.

- **Más rutas de enrutamiento** — Añadir una ruta `"critical"` para pérdida de datos o problemas de seguridad que omita la IA por completo y vaya directamente a un humano.

- **Historial del cliente** — Dar a los agentes acceso a los tickets anteriores del cliente e información de la cuenta para respuestas más personalizadas.

- **`and_()` para procesamiento paralelo** — Usar `and_()` para esperar a que múltiples agentes terminen antes de continuar. Por ejemplo, ejecutar análisis de sentimiento y clasificación en paralelo, y luego enrutar basándose en ambos resultados.

Pero la arquitectura central — un **Flow** que clasifica, enruta y orquesta **Crews** de agentes especializados — sigue siendo exactamente la misma.
